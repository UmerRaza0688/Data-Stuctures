# All and Any 

```python
a = 5
b = 4

if a > 2 and a < 10 and b < 10 and b > 0:
    print("All conditions are satisfied ):-")
```

```python
# Alternative approach, increase readability
a = 5 
a = 4
# a = 10  # to test any
constraints = { a > 2, a < 10, b < 10, b > 0 }
```

```python
if all(constraints):
    print("All conditions are satisfied ):-")
```

```python
if any(constraints):
    print("At least one condition is satisfied.")
```

# Quantifiers

```python
X = {10, 20, 30, 40, 50}
```

```python
all( [x < 100 for x in X] )      # For all x in X, x < 100
```

```python
any( [x < 50 for x in X] )       # some exists x in X, x < 100
```

```python
def div_by_20(n):
    return n % 20 == 0
```

```python
all( [div_by_20(x) for x in X] )      # For all x in X,div by 20
```

```python
any( [div_by_20(x) for x in X] )      # some exists x in X,div by 20
```

# Counter

```python
from collections import Counter
cnt = Counter()

for word in [ 'red', 'blue', 'black', 'red', 'black', 'black' ]:
    cnt[word] += 1
```

```python
cnt
```

```python
cnt['red']
```

```python
cnt.most_common(1)
```

```python
cnt.most_common()[0][0]   #most common just name
```

```python
cnt.most_common()[-1][0]   # least common just name
```

```python
from collections import Counter
cnt = Counter()

lines = """Whether the bygone melody comes back whether it does not.
            Whether a sweet breeze blows from Hijaz whether it does not.
            This mendicant’s life draws to its close, 
            Whether a knower of the secret comes again whether not"""

for word in lines.split():
    cnt[word] += 1

cnt.most_common()
```
