```python
def fib1(n):
    if n <= 1:
        return 1
        
    else:
        return fib1(n - 1) + fib1(n - 2)
```

```python
def fib2(n):
    a = 1;
    b = 1;

    for i in range(n):
        a, b = b, a+b

    return a
```

```python
print("fib1:  ")
for i in range(15):
    print(fib1(i), end=', ')

print("\n\n")
print("fib2: ")
for i in range(15):
    print(fib2(i), end=', ')
```

```python
# Checking how much time take, these two algos
```

```python
import time
```

```python
def compute_time(fn, limit):
    l = []

    for i in range(limit):

        start_time = int(round(time.time() * 1000))

        fn(i);

        end_time = int(round(time.time() * 1000))
        time_taken = end_time - start_time

        l.append(time_taken)

    return l
```

```python
limit = 30
fib1_times = compute_time(fib1, limit)
fib2_times = compute_time(fib2, limit)
```

```python
plt.figure()
plt.xlabel('fib(n)')
plt.ylabel('times(ms)')
plt.plot(fib1_times)
plt.plot(fib2_times)
plt.grid()
plt.show()
```
