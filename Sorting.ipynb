# Sorting Algorithms

# Bubble Sort

```python
def BubbleSort(arr):
    l = len(arr)
    for i in range(l - 1):
        swapped = False
        for j in range(l - i - 1):
            # Exchanging values
            if(arr[j] > arr[j + 1]):
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True      # if swap occure flage change
        # if swap is not occured break the loop
        if(swapped is False):
            break
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
BubbleSort(lst)
print(lst)
```

# Selection Sort

```python
n = 10
for i in range(n-1):
    print(i)
```

```python
def SelectionSort(arr):
    n = len(arr)

    # j will one step ahead of i 
    for i in range(n-1):  # i from 0 to index-1
        for j in range(i+1, n): # i+1 to n
            if arr[j] < arr[i]: # if j > i, swap else ignore
                arr[j], arr[i] = arr[i], arr[j]  
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
SelectionSort(lst)
print(lst)
```

# Insertion Sort

```python
def InsertionSort(lst):
    # go through all elements (except first)
    # now pick element, Call it Key
    # Each time key will be inserted in its place
    # At each itration, stuff less then i would be sorted already

    n = len(lst)

    for i in range(n):

        key = lst[i]   # hold the key

        # start comparing keys to things on its left
        # stop when less or equal value found (or we reach left end)

        j = i-1
        while j >= 0 and key < lst[j]:
            lst[j+1] = lst[j]          # shift val to right, until val come less then or equal to key
            j = j - 1;                 # after every val shift, shift j to left

        lst[j+1] = key;                # place key in free slot, (j+1 because we decreement j above)

            
```

```python
lst = [6, 1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
InsertionSort(lst)
print(lst)
```

# Quick Sort

```python
import random

def qsort(lst, first, last):
    if first >= last: return   # indexes

    i, j = first, last
    pivot = lst[random.randint(first, last)]   # selecting random

    while i <= j:   # until i & j don't cross eachother
        while lst[i] < pivot: i += 1   # for DESC order change '<' to '>'
        while lst[j] > pivot: j -= 1   # for DESC order change '>' to '<'

        if i <= j:
            lst[i], lst[j] = lst[j], lst[i]
            i, j = i+1, j-1


    qsort(lst, first, j)
    qsort(lst, i, last)
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
qsort(lst, 0, len(lst)-1)
print(lst)
```

# Sorting In Real World

# Descending Sorts

```python
def selection_sort(lst):
    n = len(lst)

    # for each element in the list (starting from the left)
    for i in range(n):
        idx = i      # find the replacement........

        # ..... in the *rest* of list
        for j in range(i+1, n):

            if lst[j] > lst[idx]:    # changed in logic for descending
                idx = j

        # swap the replacement with current element, now we have sorted
        lst[i], lst[idx] = lst[idx], lst[i]
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
selection_sort(lst)
print(lst)
```

But there's a better way!

```python
def less_then(a, b):
    return a < b
```

```python
def selection_sort(lst, compare_with):
    n = len(lst)

    # for each element in the list (starting from the left)
    for i in range(n):
        min_idx = i      # find the replacement........

        # ..... in the *rest* of list
        for j in range(i+1, n):

            if compare_with(lst[j],lst[min_idx]):    # now the comparison is out-source
                min_idx = j

        # swap the replacement with current element, now we have sorted
        lst[i], lst[min_idx] = lst[min_idx], lst[i]
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
selection_sort(lst, less_then)
print(lst)
```

```python
def greater_then(a, b):
    return a > b
```

```python
lst = [1, 2, 4, 1, 2, 5, 5, 6, 1, 110, 15]
selection_sort(lst, greater_then)
print(lst)
```

# Taking it Even Further

Now we can do all of sorts of stuff with this without making a single change 
to our selection sort code.

```python
all_tuples = [(24, 25),
               (1, 2),
               (2, 4),
               (3, 5)]
```

```python
def tuple_less_then(a, b):
    return sum(a) < sum(b)

def tuple_greater_then(a, b):
    return sum(a) > sum(b)
```

```python
print("Ascending:\t", end="")
selection_sort(all_tuples, tuple_less_then)
print(all_tuples)

print("Descending:\t", end="")
selection_sort(all_tuples, tuple_greater_then)
print(all_tuples)
```

# Sorting in Python

If you have a list of dictionries--each representing a student, for instance.

```python
dic = [
    { 'name': 'Khalid', 'age': 7 },
    { 'name': 'Umer', 'age': 12 },
    { 'name': 'Raza', 'age': 5 },
    { 'name': 'Ali', 'age': 15 }
]
```

```python
def dic_less_then(a, b):
    return a['age'] < b['age']

selection_sort(dic, dic_less_then)
print(dic)
```

```python
def student_age(a):
    return a['age']
```

```python
print(dic)
dic.sort(key=student_age, reverse=True)
print(dic)
```

# Sorting Objects of Custom Classes

```python
def student_age(a):
    return a.age
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def __str__(self):
        return self.name + ': ' + str(self.age)
```

```python
s1 = Student('Wajid', 5)
s2 = Student('Raza', 12)
s3 = Student('Umer', 17)

s = [s1, s2, s3]
```

```python
for i in s:
    print(i)
```

# Anonymous Function

```python
s.sort(key=lambda x: x.age)
```

```python
for i in s:
    print(i)
```
