# Weighted Directed Graph

```python
class WeightedDiGraph:
    def __init__(self):
        # dictionary to store the graph, each node is a key and
        # its value is a list of nodes it points to
        self.g = {}        
```

```python
# add a node to the graph, with empty list because no edges yet
def add_node(self, node):
    if node in self.g:
        raise ValueError("Node already in graph")
        
    self.g[node] = []

WeightedDiGraph.add_node = add_node
```

```python
# add edge
def add_edge(self, src, dest, weight):
    # both nodes must exist before we can connect them
    if src not in self.g:
        raise ValueError("Source node not exist in graph")
    if dest not in self.g:
        raise ValueError("Destination node not exist in graph")
    

    # get the adjacency list of src so we can add dest to it
    nexts = self.g[src]
    # if src & dest are already connected, avoid duplicate edges, directed graph doesn't need same edge twice
    if dest in nexts:
        return

    # if not connected, connect src -> dest by adding dest to src's neighbor list
    nexts.append((dest, weight))

WeightedDiGraph.add_edge = add_edge
```

```python
g = WeightedDiGraph()

# Adding nodes to graph
nodes = ['a', 'b', 'c', 'd', 'e', 'f']

for n in nodes:
    g.add_node(n)

# Adding Edges
edges = [
    ('a', 'b', 5),
    ('a', 'c', 1),
    ('b', 'c', 7),
    ('b', 'd', 8),
    ('c', 'd', 3),
    ('d', 'c', 4),
    ('e', 'f', 8),
    ('f', 'c', 7)
]

for e in edges:
    g.add_edge(e[0], e[1], e[2])
```

# Traversing Through Graph Using BFS

```python
# in graph, we have to give a starting point
def traverse_graph(self, start):
    """Traverse graph from a given starting point."""
    # sanity check
    if start not in self.g:
        raise ValueError('Given node do not exist.')

    q = [start]  # Queue, initilize with starting point
    visited = [] # you will understand in code, for what it is for

    while q:     # untile q is not empty
        current = q.pop(0)

        # if we already visited the node, then we can skip it.
        # in 1st iteration it's empty
        if current in visited:
            continue

        # perform any operation here, for simplicity we are only printing nodes
        print(current)
        
        # Now operation is performed on current node, just put it in the visited list
        visited.append(current)

        # Now get all directly connected nodes to current or neighbours of current
        next_nodes = self.g[current]

        # append them in queue
        for n in next_nodes:
            q.append(n[0])      # get node only instead go getting weight also, like (node, weight)

WeightedDiGraph.traverse_graph = traverse_graph
```

```python
g.traverse_graph('a')
```

# Shortest Path - Dijkstra's Algorithm

```python
# This function is not finding shortest path from a given source to a specific destination,
# it find shortest path from given node to the last nast node in the graph in it's range
def find_shortest_path_cost_dijkstra(self, src):
    # Mark all nodes unvisited and store them.
    to_visit = list( self.g.keys() )

    print("To visit: " + str(to_visit))

    # Set the distance to zero for our initial node and to infinity for other nodes
    inf = float('inf')  # that's python for python
    dists = { node: inf for node in to_visit }
    dists[src] = 0
    print("All distances: " + str(dists))

    # let's loop
    while to_visit:
        print("-------------")

        # select the unvisited node with the smallest distance
        # can't compare 'a' with 'b'. So, we compare Dists['a'] with dists['b']
        # by the help of anonymous function
        current = min(to_visit, key=lambda node: dists[node])
        print("Current: " + current)

        # check to make sure min distance isn't infinity
        if dists[current] == inf:
            break

        # Find unvisited neighbors for the current node
        nexts = self.g[current]
        unvisited_neighbors = []
        for n in nexts:
            if n[0] in to_visit:            # recall that n e.g ('b', 5)
                unvisited_neighbors.append(n)

        print("Unvisited neighbors of " + current + ": " + str(unvisited_neighbors))


        # calculate their distances through the current
        for n in unvisited_neighbors:
            lable = n[0]
            dist_to = n[1]

            old_distance = dists[lable]
            new_distance = dists[current] + dist_to

            # See which is better: old best distance  or the new one
            if new_distance < old_distance:
                dists[lable] = new_distance

        print("All distances" + str(dists))

        # current is now visited 
        to_visit.remove(current)

        # break                # break after each iteration for demo

WeightedDiGraph.find_shortest_path_cost_dijkstra = find_shortest_path_cost_dijkstra
```

```python
g = WeightedDiGraph()

# Adding nodes to graph
nodes = ['S', 'A', 'B', 'C', 'D', 'E', 'F', 'G']

for n in nodes:
    g.add_node(n)

# Adding Edges
edges = [
    ('S', 'A', 3),
    ('S', 'C', 4),
    ('S', 'D', 9),
    ('A', 'B', 4),
    ('B', 'E', 6),
    ('B', 'G', 8),
    ('C', 'B', 8),
    ('C', 'D', 3),       # remove this first
    ('C', 'E', 5),        #then this too
    ('D', 'E', 2),
    ('D', 'G', 14),
    ('E', 'F', 3),
    ('E', 'G', 3),
    ('F', 'B', 5)
]

for e in edges:
    g.add_edge(e[0], e[1], e[2])
```

```python
# This function is finding shortest path from a given source to a specific destination,
# with it's accumulative cost
def find_shortest_path_dijkstra(self, src, dest):
    # Mark all nodes unvisited and store them.
    to_visit = list( self.g.keys() )

    print("To visit: " + str(to_visit))

    # Set the distance to zero for our initial node and to infinity for other nodes
    inf = float('inf')  # that's python for python
    dists = { node: inf for node in to_visit }
    dists[src] = 0
    print("All distances: " + str(dists))

    # to store path in dict
    best_paths = {}
    best_paths[(src, src)] = [src] # initial path, no move

    # let's loop
    while to_visit:
        print("-------------")

        # select the unvisited node with the smallest distance
        # can't compare 'a' with 'b'. So, we compare Dists['a'] with dists['b']
        # by the help of anonymous function
        current = min(to_visit, key=lambda node: dists[node])
        print("Current: " + current)

        # check to make sure min distance isn't infinity
        if dists[current] == inf:
            break

        # Find unvisited neighbors for the current node
        nexts = self.g[current]
        unvisited_neighbors = []
        for n in nexts:
            if n[0] in to_visit:            # recall that n e.g ('b', 5)
                unvisited_neighbors.append(n)

        print("Unvisited neighbors of " + current + ": " + str(unvisited_neighbors))


        # calculate their distances through the current
        for n in unvisited_neighbors:
            lable = n[0]
            dist_to = n[1]

            # get old best distance and newdistance
            old_distance = dists[lable]
            new_distance = dists[current] + dist_to

            # See which is better: old best distance  or the new one
            if new_distance < old_distance:
                print("\nFound new best path.....")
                dists[lable] = new_distance

                # also save path with cost
                # best way from source to lable is src->current->lable
                path_to_current = best_paths[(src, current)][:] # need a copy
                best_paths[(src, lable)] = path_to_current
                best_paths[(src, lable)].append(lable)
                print("Previous best path to current: ", best_paths[(src, current)])
                print("Best path to: ", lable, ": ", best_paths[(src, lable)])
                    

        print("All distances" + str(dists))

        # current is now visited 
        to_visit.remove(current)

        # break                # break after each iteration for demo
        
    return best_paths[(src, dest)], dists[dest]

WeightedDiGraph.find_shortest_path_dijkstra = find_shortest_path_dijkstra
```
