# Factorial

```python
def fact(n):
    # base case
    if n <= 1:
        return 1

    # induction case
    return n * fact(n-1)
```

```python
for i in range(1, 7):
    print("Factorial of", i, "is: ", fact(i))
```

# Fib

```python
# Algo 1: slow and readable
def fib(n):
    # base case
    if n <= 1:
        return n

    return fib(n-1) + fib(n-2)
```

```python
%timeit fib(5)
fib(5)
```

```python
# Algo 2: by iteration, effiecent
def fib(n):
    a, b = 0, 1
    for i in range(n):
        a, b = b, a+b
    return a
```

```python
%timeit fib(5)
fib(5)
```

```python
# Algo 3: contains pros of both, effiecent but not readable
def fib(n, a=0, b=1):
    if n == 0:
        return a

    return fib(n-1, b, a+b)
```

```python
%timeit fib(5)
fib(5)
```

# Tower of Hanoi

```python
def tower_of_hanoi(levels = 3):
    move_tower(levels, 'A', 'C', 'B')

def move_tower(l, fr, to, aux):    # (l, from, to, auxlary)
    if l == 1:
        print_move(l, fr, to)
        return

    move_tower(l-1, fr, aux, to)
    print_move(l, fr, to)
    move_tower(l-1, aux, to, fr)

def print_move(l, fr, to):
    print("Move: ", l, "from", fr, "to", to)
```

```python
tower_of_hanoi(2)
```

# Sum Over a List 

```python
l = [2, 4, 5, 8]   # 19
```

```python
# by using recursion
def sum_of_list(l):
    # if list is empty, base case
    if len(l) == 0:
        return 0

    return l[0] + sum_of_list(l[1:])
```

```python
sum_of_list(l)
```

```python
# by iteration
def sum_of_list(l):
    _sum = 0

    for i in l:
        _sum += i
    return _sum
```

```python
sum_of_list(l)
```
