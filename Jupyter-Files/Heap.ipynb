# Heap Data Structure - MaxHeap and MinHeap

```python
def heapify(lst, n, root):
    """Heapify the root element of lst which has n elements in total"""

    maximum = root
    left = 2*root + 1
    right = 2*root + 2

    if left < n and lst[left] > lst[maximum]:
        maximum = left

    if right < n and lst[right] > lst[maximum]:
        maximum = right

    # if maximum is changed
    if maximum != root:
        lst[root], lst[maximum] = lst[maximum], lst[root]
        # recurssive call to manage if swapping change the analogy of heap
        heapify(lst, n, maximum)
```

```python
def build_heap(lst):
    """Construct heap from any given list by repeatedly calling heapify from bottom-to-top"""

    n = len(lst)

    for i in reversed(range(n // 2)): # in complete BST, half nodes must be the leaf nodes, so no need to heapify them
        heapify(lst, n, i)
```

```python
heap = [100, 5, 3, 2, 8, 15, 6, 102]
```

```python
print(heap)
build_heap(heap)
print(heap)
```

# Heap Sort

```python
def heap_sort(lst):
    """Perform heap sort by building a MaxHeap from the list then and then repeatedly elements"""

    n = len(lst)

    # Build Max-Heap
    build_heap(lst)


    for i in reversed(range(n)):
        # swap
        lst[i], lst[0] = lst[0], lst[i]

        # heapify root element again but only until ith element
        heapify(lst, i, 0)
```

```python
heap = [100, 5, 3, 2, 8, 15, 6, 102]
print(heap)
heap_sort(heap)
print(heap)
```
