```python
# File size
file_sizes = []

for dir_path, dirs, files in os.walk("."):
    print("Running in: ", os.path.join(dir_path))

    for f in files:
        file_full_path = os.path.join(dir_path, f)
        fileSize = os.path.getsize(file_full_path)

        file_sizes.append((file_full_path, fileSize))

    print("")
```

```python
filtered = filter(lambda x: x[1] > (1024*1024)*25, file_sizes)  #1024*1024=1MB
for f in filtered:
    print(f)
```

# Deck of Cards

```python
class Card:
    def __init__(self, suit, val):
        self.suit = suit
        self.val = val

    def __str__(self):
        return str(self.val) + ' of ' + self.suit
```

```python
# Deck is consist of 52 cards, card has seprate class
class Deck:
    def __init__(self):
        self.cards = []
        self.build()

    def build(self):
        suit = ['Spades', 'Hearts', 'Diamonds', 'Clubs']  # Shapes
        for s in suit:    # type of suit
            for v in range(1, 14):   # no of card, 1st 9 cards has # 1 to 10 and 11, 12, 13 = King, Queen, joker,  
                self.cards.append(Card(s, v))   # we will create a card and add it to Deck 1 by 1

    def __str__(self):
        ret = ""
        for c in self.cards:
            ret += str(c) + '\n'
        return ret
```

```python
c1 = Card("Hearts", 8) # single card
print(c1)
```

```python
d = Deck()
print(d)
```

```python
# Shuflling Deck of Cards
import random
```

```python
def shuffle(self):
    for i in range(0, len(self.cards)):
        r = random.randint(0, i)  # find another number
        self.cards[i], self.cards[r] = self.cards[r], self.cards[i]   # and swap

Deck.shuffle = shuffle
```

```python
d.shuffle()
print(d)
```

```python
def draw(self):
    r = random.randint(0, len(self.cards))
    c = self.cards.pop(r)
    return c

Deck.draw = draw
```

```python
c = d.draw()
```
