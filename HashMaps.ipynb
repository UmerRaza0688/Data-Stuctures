# HashMap

```python
class HashMap:
    def __init__(self):
        self.size = 10
        self.map = [None] * self.size
```

```python
def _get_hash(self, key):
    return key % self.size

HashMap._get_hash = _get_hash
```

```python
def add(self, key, value):
    key_hash = self._get_hash(key)
    key_value = [key, value]

    # insert or update, in hash table
    
    self.map[key_hash] = [ key_value ]

HashMap.add = add
```

```python
def get(self, key):
    key_hash = self._get_hash(key)
    if self.map[key_hash] is not None:
        for pair in self.map[key_hash]:
            if pair[0] == key:
                return pair[1]

    raise KeyError(str(key))

HashMap.get = get
```

```python
def remove(self, key):
    key_hash = self._get_hash(key)
    if key_hash is not None:
        for pair in self.map[key_hash]:
            if pair[0] == key:
                self.map[key_hash].pop()
                return True
            return False

HashMap.remove =remove
```

```python
# change value at given key
def update(self, key, value):
    key_hash = self._get_hash(key)
    if self.map[key_hash] is not None:
        for pair in self.map[key_hash]:
            if pair[0] == key:
                pair[1] = value
                return True
            return False

            
HashMap.update = update
```

```python
# to understand enumerate
lst = [ 'a', 'b', 'c' ]
for i, item in enumerate(lst):
    print(i, '----', item)
```

```python
def __str__(self):
    ret = ""

    for i, item in enumerate(self.map):
        if item is not None:
            ret += str(i) + ": " + str(item) + "\n"

    return ret

HashMap.__str__ = __str__
```

```python
h = HashMap()
```

```python
h.add(13, "thirteen")
h.add(15, "fifteen")
h.add(27, "twenty seven")
h.add(13, "thirteen-----repeated")  # key cannot be same in a hashTable, so it will overwrite
h.add(25, "twenty five")  # after adding key 25, where is 15? This is called collision, a problem in hashtables.

print(h)

# h.get(15)  # key error
h.get(25)

h.remove(25)
print(h)

h.update(27, "stais")
print(h)
```

# Collisions and Avoidance

```python
hm = HashMap()

hm.add(17, "seventeen")
hm.add(26, "twenty six")
hm.add(35, "thirty five")
hm.add(25, "twenty five")
hm.add(26, "twenty six updated")
hm.add(887, "large number, with repeated mod val")

print(hm)
```

```python
def add(self, key, value):
    key_hash = self._get_hash(key)
    key_value = [key, value]

    if self.map[key_hash] is None:
        self.map[key_hash] = [key_value]
        return True

    else:
        # check if it's an update
        for pair in self.map[key_hash]:
            if pair[0] == key:
                print("Updating: ", key)
                pair[1] = value
                return True

    # if it's a collision, insert in in the list, same values are handeled
    self.map[key_hash].append(key_value)
    return True

HashMap.add = add
```

```python
hm = HashMap()

hm.add(17, "seventeen")
hm.add(26, "twenty six")
hm.add(35, "thirty five")
hm.add(25, "twenty five")
hm.add(26, "twenty six updated")
hm.add(887, "large number, with repeated mod val")

print(hm)
```

```python
def delete(self, key):
    key_hash = self._get_hash(key)

    if self.map[key_hash] is None:
        print("KeyError!")
        return

    for i in range(0, len(self.map[key_hash])):
        if self.map[key_hash][i][0] == key:
            self.map[key_hash].pop(i)
            return True

HashMap.delete = delete
```
