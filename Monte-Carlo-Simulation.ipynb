# Monte Carlo Simulation

We can do simulations to, well, simulate the real world in a low-risk 
enviroment. Monte Carlo Simulation is by far the most widely used simulation
and the concept is very simple: try to do something n number of times and calculate
probabilities of landing in a particular size.

Examples:

1. Estimating Pi
2. Random Walks
3. Finance: Predicting stock prices
4. Aeronautical Engineering: Airflow around the wings
5. Car Safety: Car simulations
6. Medicine: Spread of disease and contagions
7. Much more...........

# Estimating Pi

Area of circle is given as: $\pi r^2$

So, $pi$ is essentially a ratio b/w the area and the radius of square. We can use that to
estimate the value of $pi$ using Monto-Carlo-Simulation.

We can simulate throwing darts and see which ones land within the circle and which one land
outside it.

Also, if we throw darts only in the top-right quadrant. It will have the same ratio.

Area of circle = $\pi r^2$
Area of square = $(2r)^2$ = $4 r^2$

So if we divide the area of circle by the area of square, we get $\pi/4$. Multiply by 4 and you get
an estimate of the value of the $\pi$.


```python
import random as rand
from math import sqrt
```

```python
def distance(a, b):
    return sqrt(a**2 + b**2)

def estimate_pi(darts):
    inside = 0      # initially darts in the circle

    for i in range(darts):
        x = rand.random()
        y = rand.random()

        if distance(x, y) < 1.0:
            inside += 1

    pi = (inside / darts) * 4
    return pi
```

```python
num_sim = int(1e+6)
pi = estimate_pi(num_sim)
print(pi)
```

# Random Walks

```python
steps = 10000

import random

def get_action():
    return random.choice( [ (0, 1), (0, -1), (1, 0), (-1, 0) ] )

def random_walk(n):
    """ Return path after 'n' block random walk"""
    path = []

    x, y = 0, 0

    for i in range(n):

        (dx, dy) = get_action()     # get step deltas

        x = x+dx
        y = y+dy

        path.append( (x, y) )      # create new tuple and save in the 'path'

    return path


walk = random_walk(steps)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.figure(num=None, figsize=(15, 10))


point_color = list(range(steps))

x_numbers = [x[0] for x in walk]
y_numbers = [y[1] for y in walk]

plt.scatter(x_numbers, y_numbers, marker='.', s=50, c=point_color)

plt.show()
```
