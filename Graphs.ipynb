# Graph Visualizer

```python
from graphviz import Digraph

def visualize_graph(graph):
    if not graph.g:
        return 'Nothing in the graph!'
    
    dot = Digraph()
    dot.attr('node', shape='circle')
    
    # track visited nodes to avoid infinite loop in case of cycles
    visited = set()
    
    def add_nodes_edges(node):
        if node in visited:
            return
        visited.add(node)
        
        dot.node(name=str(node), label=str(node))
        
        for neighbor in graph.g[node]:
            dot.node(name=str(neighbor), label=str(neighbor))
            dot.edge(str(node), str(neighbor))
            add_nodes_edges(neighbor)  # recursive call for each neighbor
    
    # start from every node, in case graph is disconnected
    for node in graph.g:
        add_nodes_edges(node)
    
    display(dot)
```

```python
def print_graph(graph):
    for node in graph.g:
        # get the neighbor list for current node
        neighbors = graph.g[node]
        
        # if no neighbors, show a clear message instead of empty
        if neighbors:
            print(f"{node} -> {', '.join(neighbors)}")
        else:
            print(f"{node} -> (no outgoing edges)")
```

# Graph

```python
class DiGraph:
    def __init__(self):
        # dictionary to store the graph, each node is a key and
        # its value is a list of nodes it points to
        self.g = {}

    def __str__(self):
        result = ""
        for node, neighbors in self.g.items():
            # format: A -> B, C or A -> (no outgoing edges)
            result += f"{node} -> {', '.join(neighbors) if neighbors else '(no outgoing edges)'}\n"
        return result
```

```python
# add a node to the graph, with empty list because no edges yet
def add_node(self, node):
    if node in self.g:
        raise ValueError("Node already in graph")
        
    self.g[node] = []

DiGraph.add_node = add_node
```

```python
# add edge
def add_edge(self, src, dest):
    # both nodes must exist before we can connect them
    if src not in self.g:
        raise ValueError("Source node not exist in graph")
    if dest not in self.g:
        raise ValueError("Destination node not exist in graph")
    

    # get the adjacency list of src so we can add dest to it
    nexts = self.g[src]

    # if src & dest are already connected, avoid duplicate edges, directed graph doesn't need same edge twice
    if dest in nexts:
        return

    # if not connected, connect src -> dest by adding dest to src's neighbor list
    nexts.append(dest)

DiGraph.add_edge = add_edge
```

```python
g1 = DiGraph()

# Add nodes
nodes = ['A', 'B', 'C', 'D', 'E', 'F']  # put nodes in list

# run loop on it
for i in nodes:
    g1.add_node(i)
```

# Traversing Through Graph Using BFS

```python
# in graph, we have to give a starting point
def traverse(self, start):
    """Traverse graph from a given starting point."""
    # sanity check
    if start not in self.g:
        raise ValueError('Given node do not exist.')

    q = [start]  # Queue, initilize with starting point
    visited = [] # you will understand in code, for what it is for

    while q:     # untile q is not empty
        current = q.pop(0)

        # if we already visited the node, then we can skip it.
        # in 1st iteration it's empty
        if current in visited:
            continue

        # perform any operation here, for simplicity we are only printing nodes
        print(current)
        
        # Now operation is performed on current node, just put it in the visited list
        visited.append(current)

        # Now get all directly connected nodes to current or neighbours of current
        next_nodes = self.g[current]

        # append them in queue
        for n in next_nodes:
            q.append(n)

DiGraph.traverse = traverse
```

# Find Path

```python
def find_path(self, start, end, path=[]):
    """Find path ( not necessariy shortest ) from start to end """
    
    # sanity Checks
    if start not in self.g:
        raise ValueError('Source node do not exist!')
    if end not in self.g:
        raise ValueError('Destination node do not exist!')

    # print path
    print(start, ' -> ', end) 

    # starting point must be included in path
    # path.append(start)
    path = path + [start]

    # base case
    if start == end:
        return path

    # recusive cases
    for node in self.g[start]:  # check every node connected to our starting node

        # case, to avoid cycles
        if node not in path:
            # new_path is a list, because at the end function will return a list or None
            new_path = self.find_path(node, end, path) # find path from next node to end
            # if path founded, return it
            if new_path:
                return new_path

    # if no path can be found from any of the next nodes to the end, there's no path
    return None

DiGraph.find_path = find_path
```

# Find All Paths

```python
def find_all_paths(self, start, end, path=[]):
    """Function for finding all path from source to destination."""

    # Sanity Checks
    if start not in self.g:
        raise ValueError('Source node do not exist.')
    if end not in self.g:
        raise ValueError('Destination node do not exist.')

    # save the path we have traversed till now
    path = path + [start]

    # base case
    if start == end:
        return [ path ]          # return the path in a list, since we may many paths

    all_paths = []
    
    # recusive case
    for node in self.g[start]:

        # need to avoid cycles
        if node not in path:

            # find path from next node, But here we will save every path instead of returning the path
            all_newpaths = self.find_all_paths(node, end, path)
            for newpath in all_newpaths:
                all_paths.append(newpath)

    # return list of all paths, every path is also list..... so list of lists
    return all_paths

DiGraph.find_all_paths = find_all_paths
```

# Find the Shortest Path

```python
def find_shortest_path(self, start, end, path=[]):
    """Function for finding all path from source to destination."""

    # Sanity Checks
    if start not in self.g:
        raise ValueError('Source node do not exist.')
    if end not in self.g:
        raise ValueError('Destination node do not exist.')

    # save the path we have traversed till now
    path = path + [start]

    # base case
    if start == end:
        return [ path ]          # return the path in a list, since we may many paths

    shortest = None              # change
    # recusive case
    for node in self.g[start]:

        # need to avoid cycles
        if node not in path:

            # find path from next node, compare new path every time & if it's len < shortest
            # new shortest = newpath
            # first exception, as we initilized shortest as None, so first time whichever the new path,
            # it will become shortest
            newpath = self.find_path(node, end, path)
            if newpath:
                if shortest is None or len(newpath) < len(shortest):
                    shortest = newpath

    return shortest

DiGraph.find_shortest_path = find_shortest_path
```
